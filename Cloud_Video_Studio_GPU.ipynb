```
# @title 3. Cấu hình & Chạy Render Video Tự Động trên Cloud
VIDEO_SOURCE = 'https://www.youtube.com/watch?v=dQw4w9WgXcQ' # @param {type:"string"}
TRIM_START = '00:00:00' # @param {type:"string"}
TRIM_END = '00:00:30' # @param {type:"string"}
REMOVE_SILENCE = True # @param {type:"boolean"}
ASPECT_MODE = '9:16-blur' # @param ["original", "9:16-blur", "9:16-crop"]
SEO_KEYWORD = 'hướng dẫn edit video tự động' # @param {type:"string"}

!curl -sL https://raw.githubusercontent.com/dung11122d/cloud-video-studio/main/scripts/cloud_editor.py -o cloud_editor.py
!python cloud_editor.py \
  --source "$VIDEO_SOURCE" \
  --trim-start "$TRIM_START" \
  --trim-end "$TRIM_END" \
  --remove-silence "$REMOVE_SILENCE" \
  --aspect "$ASPECT_MODE" \
  --seo-keyword "$SEO_KEYWORD" \
  --output-dir "output"
```
